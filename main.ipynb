```python
import numpy as np
from keras.datasets import fashion_mnist
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score
```

```python
# load and validate dataset
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
assert x_train.shape == (60000, 28, 28)
assert x_test.shape == (10000, 28, 28)
assert y_train.shape == (60000,)
assert y_test.shape == (10000,)

# reshape data to pass into models (they require 2d arrays of shape "n_samples x n_features")
x_train = x_train.reshape(60000, 784)
x_test = x_test.reshape(10000, 784)
```

```python
def pipeline(model, x_training: np.ndarray) -> tuple:
    '''
    Trains and evaluates model.

    inputs:
    model: knn, mlp, or logistic regression model
    x_training: training features

    output:
    tuple containing accuracy, precision, and recall for model.
    '''
    #train
    model.fit(x_training, y_train)
    # get predictions
    prediction = model.predict(x_test)

    # evaluate
    accuracy = model.score(x_test, y_test)
    precision = precision_score(y_test, prediction, average='macro')
    recall = recall_score(y_test, prediction, average='macro')

    return (accuracy, precision, recall)

def add_gaussian_noise(x_training: np.ndarray, scale: int) -> np.ndarray:
    '''
    Adds random values to the existing pixel intensities.

    inputs:
    x_train: flattened training data.
    scale: standard deviation of the random distribution. Higher values results in noiser data.

    output:
    noisy training data.
    '''
    noise = np.random.normal(0, scale, x_train.shape)
    return np.clip(x_train + noise, 0, 255)
    
    
```

```python
############################# kNN example (without noise) ###############################################
knn = KNeighborsClassifier(n_neighbors=5)

res = pipeline(knn, x_train)

print(
f'''
(No noise)
Model: kNN (k = {knn.n_neighbors})
Accuracy: {res[0]:.2%}
Precision: {res[1]:.2%}
Recall: {res[2]:.2%}
'''
)
```

```python
############################# kNN example (with noise) ###############################################
knn = KNeighborsClassifier(n_neighbors=5)

x_noisy = add_gaussian_noise(x_train, 100)
res = pipeline(knn, x_noisy)

print(
f'''
(Noise)
Model: kNN (k = {knn.n_neighbors})
Accuracy: {res[0]:.2%}
Precision: {res[1]:.2%}
Recall: {res[2]:.2%}
'''
)
```

```python
############################# MLP example (without noise) ###############################################
mlp = MLPClassifier(verbose=True) # defaults to 1 hidden layer with 100 neurons
res = pipeline(mlp, x_train)

print(
f'''
(No noise)
Model: MLP (layers = {len(mlp.hidden_layer_sizes)})
Accuracy: {res[0]:.2%}
Precision: {res[1]:.2%}
Recall: {res[2]:.2%}
'''
)
```

```python
############################# MLP example (with noise) ###############################################
mlp = MLPClassifier(verbose=True) # defaults to 1 hidden layer with 100 neurons
x_noisy = add_gaussian_noise(x_train, 100)
res = pipeline(mlp, x_noisy)

print(
f'''
(Noise)
Model: MLP (layers = {len(mlp.hidden_layer_sizes)})
Accuracy: {res[0]:.2%}
Precision: {res[1]:.2%}
Recall: {res[2]:.2%}
'''
)
```
